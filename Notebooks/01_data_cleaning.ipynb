## Parent and Child Tables

Parent tables give ID - movies gives movie_id, genres gives genre_id
Child tables use that ID - cast, crew, movie_genres, movie_keywords

### What is Orphan?

Orphan means child row has movie_id, but that movie_id is not there in movies table. So it has no parent.

### Why Orphan Check?

We check this before SQL, because if orphan is there, SQL will give Foreign Key error. So we remove them.

### Orphan Check - movie_keywords

We check if every movie_id in movie_keywords is there in movies table or not.

```python
orphan_keywords = movie_keywords[~movie_keywords['movie_id'].isin(movies['movie_id'])]
print("Orphan rows in movie_keywords:", len(orphan_keywords))
```

```python
print(orphan_keywords.head())
```

Observation: Orphan rows in movie_keywords = 0. So no orphan rows. Data is clean and ready for SQL.

### Orphan Check - cast

We check if every movie_id in cast is there in movies table or not.

```python
orphan_cast = cast[~cast['movie_id'].isin(movies['movie_id'])]
print("Orphan rows in cast:", len(orphan_cast))
print(orphan_cast.head())
```

Observation: Orphan rows in cast = 0. So no orphan rows. Data is clean and ready for SQL.

### Orphan Check - crew

We check if every movie_id in crew is there in movies table or not.

```python
orphan_crew = crew[~crew['movie_id'].isin(movies['movie_id'])]
print("Orphan rows in crew:", len(orphan_crew))
print(orphan_crew.head())
```

Observation: Orphan rows in crew = 0. So no orphan rows. Data is clean and ready for SQL.

### Orphan Check - movie_genres

We check if every movie_id in movie_genres is there in movies table or not.

```python
orphan_genres = movie_genres[~movie_genres['movie_id'].isin(movies['movie_id'])]
print("Orphan rows in movie_genres:", len(orphan_genres))
print(orphan_genres.head())
```

Observation: Orphan rows in movie_genres = 0. So no orphan rows. Data is clean and ready for SQL.

```python
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:Kanna0823@localhost:3306/tmdb")

movies.to_sql('movies', con=engine, if_exists='replace', index=False)
genres.to_sql('genres', con=engine, if_exists='replace', index=False)
movie_genres.to_sql('movie_genres', con=engine, if_exists='replace', index=False)
cast.to_sql('cast', con=engine, if_exists='replace', index=False)
crew.to_sql('crew', con=engine, if_exists='replace', index=False)
movie_keywords.to_sql('movie_keywords', con=engine, if_exists='replace', index=False)

print("All 6 tables loaded!")
```

```python
query = """
SELECT 'movies' AS table_name, COUNT(*) AS rows_count FROM movies
UNION ALL
SELECT 'genres', COUNT(*) FROM genres
UNION ALL
SELECT 'movie_genres', COUNT(*) FROM movie_genres
UNION ALL
SELECT 'cast', COUNT(*) FROM cast
UNION ALL
SELECT 'crew', COUNT(*) FROM crew
UNION ALL
SELECT 'movie_keywords', COUNT(*) FROM movie_keywords
"""

pd.read_sql(query, engine)
```
